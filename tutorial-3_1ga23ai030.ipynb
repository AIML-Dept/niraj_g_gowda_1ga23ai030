```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator
from IPython.display import display
```

```python
# Pauli-X Gate

qc_x = QuantumCircuit(1)

qc_x.x(0)

print("Pauli-X Gate Circuit:")
display(qc_x.draw())

print("Statevector:")
print(Statevector(qc_x))
```

```python
# Pauli-Y Gate

qc_y = QuantumCircuit(1)

qc_y.y(0)

print("Pauli-Y Gate Circuit:")
display(qc_y.draw())

print("Statevector:")
print(Statevector(qc_y))
```

```python
# Pauli-Z Gate

qc_z = QuantumCircuit(1)

qc_z.z(0)

print("Pauli-Z Gate Circuit:")
display(qc_z.draw())

print("Statevector:")
print(Statevector(qc_z))
```

```python
# H followed by Z gate

qc_hz = QuantumCircuit(1)

qc_hz.h(0)
qc_hz.z(0)

print("H-Z Circuit:")
display(qc_hz.draw())

print("Final State:")
print(Statevector(qc_hz))
```

```python
# Five gate sequence

qc_random = QuantumCircuit(1)

qc_random.h(0)
qc_random.x(0)
qc_random.z(0)
qc_random.y(0)
qc_random.h(0)

print("Five Gate Circuit:")
display(qc_random.draw())

print("Final State:")
print(Statevector(qc_random))
```

```python
# Quantum Coin Flip

coin = QuantumCircuit(1)

coin.h(0)

print("Quantum Coin Flip Circuit:")
display(coin.draw())

state = Statevector(coin)

print("State:")
print(state)
```

```python
import numpy as np

state = Statevector(coin)

probabilities = np.abs(state.data)**2

print("Heads Probability:", probabilities[0])
print("Tails Probability:", probabilities[1])
```

```python
# HZH circuit

hz_h = QuantumCircuit(1)

hz_h.h(0)
hz_h.z(0)
hz_h.h(0)


print("HZH Circuit:")
display(hz_h.draw())

print("HZH Matrix:")
print(Operator(hz_h))
```

```python
# X circuit

x_gate = QuantumCircuit(1)

x_gate.x(0)


print("X Circuit:")
display(x_gate.draw())

print("X Matrix:")
print(Operator(x_gate))
```

```python
matrix1 = Operator(hz_h).data
matrix2 = Operator(x_gate).data

if np.allclose(matrix1, matrix2):
    print("HZH is equivalent to X gate")
else:
    print("HZH and X are different")
```
