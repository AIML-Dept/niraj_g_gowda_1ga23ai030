## Exercises

```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector, plot_state_qsphere
import matplotlib.pyplot as plt
```

## Easy: |0> apply X gate

```python
qc=QuantumCircuit(1)
qc.x(0)
sv=Statevector.from_instruction(qc)
print(sv)
display(plot_bloch_multivector(sv))
```

## Medium: H then S

```python
qc=QuantumCircuit(1)
qc.h(0)
qc.s(0)
sv=Statevector.from_instruction(qc)
print(sv)
display(plot_bloch_multivector(sv))
```

## Hard: 2-qubit equal superposition

```python
qc=QuantumCircuit(2)
qc.h([0,1])
sv=Statevector.from_instruction(qc)
print(sv)
print("Normalization:",np.sum(np.abs(sv.data)**2))
display(plot_state_qsphere(sv))
```

## Real World: Small phase error

```python
theta=0.2
qc=QuantumCircuit(1)
qc.h(0)
qc.p(theta,0)
sv=Statevector.from_instruction(qc)
display(plot_bloch_multivector(sv))
```

## Reflection Answers

1. The sum of squared amplitudes equals 1 because total probability must always be 1.

2. The Bloch sphere represents only a single qubit. Multi-qubit entangled states require a higher-dimensional representation, so they cannot be fully visualized on a Bloch sphere.
